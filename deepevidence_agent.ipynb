````python
import gzip
import time
import xml.etree.ElementTree as ET

from biodsa.tools.protein.xml_index import (
    uniprot_xml_entry_to_dict,
    NS_URI,
)

def local_name(tag):
    return tag.rsplit("}", 1)[-1]


def inspect_xml_entries(xml_path, count=3, max_bytes=8 * 1024 * 1024):
    parser = ET.XMLPullParser(events=("start", "end"))

    bytes_read = 0
    found = 0
    root_seen = False
    start = time.perf_counter()

    with gzip.open(xml_path, "rb") as handle:
        while bytes_read < max_bytes:
            chunk = handle.read(
                min(64 * 1024, max_bytes - bytes_read)
            )

            if not chunk:
                break

            bytes_read += len(chunk)
            parser.feed(chunk)

            for event, element in parser.read_events():
                if event == "start" and not root_seen:
                    root_seen = True
                    print("实际根标签:", repr(element.tag), flush=True)

                if event != "end" or local_name(element.tag) != "entry":
                    continue

                found += 1

                print(
                    f"\n找到第 {found} 条，"
                    f"已读取 {bytes_read / 1024:.1f} KiB，"
                    f"耗时 {time.perf_counter() - start:.2f} 秒",
                    flush=True,
                )
                print("实际 entry 标签:", repr(element.tag), flush=True)

                expected_tag = f"{{{NS_URI}}}entry"

                if element.tag != expected_tag:
                    print(
                        "命名空间与现有解析器不一致，需要修改 NS_URI。",
                        flush=True,
                    )
                    return

                annotation = uniprot_xml_entry_to_dict(element)

                print("Accession:", annotation["accession"], flush=True)
                print("Name:", annotation["protein_name"], flush=True)
                print("Length:", annotation["sequence_length"], flush=True)
                print("Function:", annotation["function"][:1], flush=True)

                # 不打印完整 raw XML，避免 Notebook 输出过大
                print(
                    "Reaction equations:",
                    [
                        reaction["equation"]
                        for reaction in annotation["catalytic_activity"][:2]
                    ],
                    flush=True,
                )

                element.clear()

                if found >= count:
                    return

    print(
        f"\n测试结束：读取 {bytes_read / 1024**2:.2f} MiB，"
        f"找到 {found} 条 entry。",
        flush=True,
    )

    if found == 0:
        print("没有找到 entry，请检查文件内容或 XML 格式。")


inspect_xml_entries(str(xml_path))
````

````python
import sqlite3
import pandas as pd

DB_PATH = "/home/data4/xiangyu/dataset/chemistry_knowledgebase/ORD/ord-data/ord_nlp_lite.db"

def main():
    # 连接到刚才生成的精简版数据库
    conn = sqlite3.connect(DB_PATH)

    print("==================================================")
    print(" 验证 1：基本数据量统计")
    print("==================================================")
    
    total_rxns = pd.read_sql("SELECT COUNT(*) as count FROM reaction_conditions", conn).iloc[0]['count']
    total_comps = pd.read_sql("SELECT COUNT(*) as count FROM reaction_components", conn).iloc[0]['count']
    
    print(f"总反应数量 (reaction_conditions): {total_rxns:,}")
    print(f"总组件数量 (reaction_components): {total_comps:,}\n")

    print("==================================================")
    print(" 验证 2：产率预测任务 (Yield Prediction)")
    print(" (目标：需要反应SMILES和产率)")
    print("==================================================")
    
    # 随机抽取 3 条带有产率的反应
    query_yield = """
        SELECT reaction_id, reaction_smiles, yield_percent, temperature_celsius
        FROM reaction_conditions 
        WHERE yield_percent IS NOT NULL AND reaction_smiles IS NOT NULL
        ORDER BY RANDOM() LIMIT 3
    """
    df_yield = pd.read_sql(query_yield, conn)
    for _, row in df_yield.iterrows():
        print(f"[ID]: {row['reaction_id']}")
        print(f"[SMILES]: {row['reaction_smiles']}")
        print(f"[Yield]: {row['yield_percent']}% | [Temp]: {row['temperature_celsius']}°C\n")


    print("==================================================")
    print(" 验证 3：溶剂推断任务 (Solvent Prediction)")
    print(" (目标：需要反应SMILES和对应的溶剂集合)")
    print("==================================================")
    
    # 随机抽取 3 条记录了明确溶剂的反应
    query_solvent = """
        SELECT reaction_id, reaction_smiles, solvents 
        FROM reaction_conditions 
        WHERE solvents IS NOT NULL 
        ORDER BY RANDOM() LIMIT 3
    """
    df_solvent = pd.read_sql(query_solvent, conn)
    for _, row in df_solvent.iterrows():
        print(f"[ID]: {row['reaction_id']}")
        print(f"[SMILES]: {row['reaction_smiles']}")
        print(f"[Solvents]: {row['solvents']}\n")


    print("==================================================")
    print(" 验证 4：角色分布统计与强制包含 Reagent 的反应")
    print("==================================================")
    
    # 1. 看看整个数据库里各种 Role 的数量分布
    query_roles = """
        SELECT reaction_role, COUNT(*) as count 
        FROM reaction_components 
        GROUP BY reaction_role
        ORDER BY count DESC
    """
    df_roles = pd.read_sql(query_roles, conn)
    print("【全局组件角色分布】:")
    print(df_roles.to_markdown(index=False))
    print("\n")

    # 2. 强制随机找一个明确打上了 'REAGENT' 标签的反应
    query_target = """
        SELECT reaction_id FROM reaction_components 
        WHERE reaction_role = 'REAGENT'
        LIMIT 1
    """
    target_id = pd.read_sql(query_target, conn).iloc[0]['reaction_id']
    
    query_components = f"""
        SELECT component_type, reaction_role, compound_name, smiles 
        FROM reaction_components 
        WHERE reaction_id = '{target_id}'
    """
    df_components = pd.read_sql(query_components, conn)
    
    print(f"【明确包含 REAGENT 的反应】 ID: {target_id}")
    print(df_components.to_markdown(index=False))
    print("\n==================================================")

    conn.close()

if __name__ == "__main__":
    main()
````

````python
from rdkit import Chem
from rdkit.Chem import Descriptors
from rdkit.Chem import Crippen, Descriptors, Lipinski
import json

def predict_esol(smiles: str) -> str:
    """
    Delaney ESOL aqueous solubility prediction.

    Returns:
        {"answer": <predicted LogS>}
    """

    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        return json.dumps({"answer": "Invalid SMILES"})

    # Molecular weight
    mw = Descriptors.MolWt(mol)

    # Wildman-Crippen logP
    logp = Crippen.MolLogP(mol)

    # Rotatable bonds
    rotors = Lipinski.NumRotatableBonds(mol)

    # Aromatic proportion
    heavy_atoms = mol.GetNumHeavyAtoms()

    aromatic_atoms = sum(
        atom.GetIsAromatic()
        for atom in mol.GetAtoms()
    )

    ap = aromatic_atoms / heavy_atoms if heavy_atoms else 0.0

    # Delaney ESOL
    #
    # logS =
    # 0.16
    # - 1.5 * logP
    # - 0.01 * (MW - 40)
    # + 0.066 * RB
    # + 0.066 * AP
    #
    # NOTE:
    # Use the exact implementation/coefficients required
    # by the benchmark if its GT was generated by a
    # particular ESOL implementation.

    logS = (
        0.16
        - 1.5 * logp
        - 0.01 * (mw - 40.0)
        + 0.066 * rotors
        + 0.066 * ap
    )

    return json.dumps({
        "answer": round(logS, 3)
    })

# 测试你题目中的分子 (雌酮)
print(predict_esol("CCCCNC(=O)N1C(NC(=O)OC)=NC2=CC=CC=C21"))
# 输出: {"answer": -3.816}
````

````python
import selfies as sf
from rdkit import Chem
from rdkit.Chem import AllChem
from pyscf import gto, scf

def calculate_lumo_from_selfies(selfies_str):
    # 1. 格式转换：SELFIES -> SMILES
    smiles = sf.decoder(selfies_str)
    
    # 2. RDKit 生成 3D 坐标
    mol = Chem.MolFromSmiles(smiles)
    mol = Chem.AddHs(mol) # 必须加氢原子才能进行量子计算
    
    # 嵌入 3D 坐标并进行分子力场初步优化 (MMFF94)
    AllChem.EmbedMolecule(mol, randomSeed=42)
    AllChem.MMFFOptimizeMolecule(mol)
    
    # 3. 将 RDKit 的坐标提取为 PySCF 能读懂的格式
    xyz = []
    for atom in mol.GetAtoms():
        pos = mol.GetConformer().GetAtomPosition(atom.GetIdx())
        xyz.append(f"{atom.GetSymbol()} {pos.x:.6f} {pos.y:.6f} {pos.z:.6f}")
    pyscf_geom = "\n".join(xyz)
    
    # 4. PySCF 密度泛函理论 (DFT) 计算
    # 这里我们使用 B3LYP 泛函和 6-31G* 基组 (与 QM9 类似)
    print("开始运行 DFT 计算，请稍候...")
    pyscf_mol = gto.M(atom=pyscf_geom, basis='6-31g*', verbose=0)
    mf = scf.RKS(pyscf_mol)
    mf.xc = 'b3lyp'
    mf.kernel() # 执行自洽场计算
    
    # 5. 提取 LUMO 能量
    # 分子轨道按能量排序，寻找刚好没有电子填充的那个轨道
    lumo_idx = pyscf_mol.nelectron // 2  
    lumo_energy = mf.mo_energy[lumo_idx] # 单位: Hartree
    
    return lumo_energy

# 测试你提问过的分子
test_selfies = "[C][C][=Branch1][C][=O][N][C][=Branch1][C][=O][O][C][O][Ring1][=Branch2]"
print("计算出的 LUMO (Hartree):", calculate_lumo_from_selfies(test_selfies))
````

````python
import re

STOPWORDS = {
    "a", "an", "the", "is", "are", "was", "were", "be", "been",
    "being", "of", "for", "to", "in", "on", "at", "by", "with",
    "and", "or", "as", "it", "its", "this", "that", "these",
    "those", "which", "who", "both", "from", "into", "their",
    "acts", "act", "plays", "play", "role",
}

# 可持续扩充的常见生物医学同义表达
SYNONYMS = {
    "metallopeptidase": "metalloprotease",
    "metallopeptidases": "metalloprotease",
    "metalloproteases": "metalloprotease",
}


def normalize(text: str) -> str:
    text = text.lower()

    # 统一常见格式
    text = re.sub(r"\batp[-\s]+dependent\b", "atp dependent", text)
    text = re.sub(r"\bzinc[-\s]+dependent\b", "zinc dependent", text)

    # 统一已知同义词
    for source, target in SYNONYMS.items():
        text = re.sub(
            rf"\b{re.escape(source)}\b",
            target,
            text,
        )

    return text


def content_tokens(text: str) -> set[str]:
    tokens = re.findall(r"\b[a-z0-9]+\b", normalize(text))

    result = set()
    for token in tokens:
        if token in STOPWORDS or len(token) <= 1:
            continue

        # 简单处理名词复数
        if token.endswith("ies") and len(token) > 4:
            token = token[:-3] + "y"
        elif token.endswith("s") and len(token) > 4 and not token.endswith("ss"):
            token = token[:-1]

        result.add(token)

    return result


def protein_reward(prediction: str, reference: str,
                   threshold: float = 0.85) -> dict:
    """Binary reward with reference-content coverage diagnostics."""

    if not isinstance(reference, str) or not reference.strip():
        return {"reward": 0, "coverage": 0.0, "status": "invalid_gt"}

    if not isinstance(prediction, str) or not prediction.strip():
        return {"reward": 0, "coverage": 0.0, "status": "empty_prediction"}

    pred_norm = normalize(prediction)
    ref_norm = normalize(reference)

    # GT 原文完整出现，直接满分
    if ref_norm in pred_norm:
        return {"reward": 1, "coverage": 1.0, "status": "exact_match"}

    gt_tokens = content_tokens(reference)
    pred_tokens = content_tokens(prediction)

    if not gt_tokens:
        return {"reward": 0, "coverage": 0.0, "status": "empty_gt_tokens"}

    matched = gt_tokens & pred_tokens
    coverage = len(matched) / len(gt_tokens)

    return {
        "reward": int(coverage >= threshold),
        "coverage": round(coverage, 4),
        "status": "pass" if coverage >= threshold else "fail",
        "missing_tokens": sorted(gt_tokens - pred_tokens),
    }

prediction="## Protein feature summary  ### Overall identification The 652-aa sequence is **strongly supported as an FtsH-family ATP-dependent zinc metalloprotease**. Its closest reported reviewed homolog is *Cupriavidus metallidurans* FtsH (UniProt **Q1LLA9**): **62.3% sequence identity**, 92.8% query coverage, and E-value 3.43 × 10⁻²⁶⁸. A *Salmonella* FtsH homolog also shows about 61% identity across most of the sequence. This degree of full-length similarity supports the family assignment, though it is not direct experimental characterization of this exact sequence.  ### Sequence-derived properties | Feature | Estimate | |---|---:| | Length | 652 amino acids | | Molecular mass | 71,515 Da | | Theoretical pI | 5.81 | | GRAVY | −0.176 | | Instability index | 32.49 |  These are sequence-based estimates; they do not account for processing, modifications, ligands, or assembly state.  ### Predicted architecture and likely function - **N-terminal membrane anchor:** The sequence contains two strong hydrophobic segments, approximately around residues **17–39** and **116–139**. This fits the two-pass inner-membrane topology annotated for the close FtsH homologs. The intervening region is expected to face the periplasm; most of the protein after the second membrane span is expected to be cytosolic. These boundaries are approximate, not a topology prediction validated for this query. - **Large cytosolic catalytic region:** The sequence contains an ATPase P-loop-like motif (...GPNGCGKT...) and a downstream metalloprotease-like region. Together with the close FtsH homology, this supports the typical FtsH architecture: an ATPase motor coupled to a zinc-dependent protease. - **Likely role:** Processive degradation of cytoplasmic and membrane proteins, including membrane-protein quality control. ...  ### Limitations The evidence supports a high-confidence **FtsH-family** assignment, but specific substrates, physiological roles, and regulation cannot be assigned from this sequence comparison alone. Pfam scanning was unavailable because the required database was not configured, and the evidence-graph retrieval did not return the recorded entries."
reference="A brief overview of the protein with the provided amino acid sequence is as follows: Acts as a processive, ATP-dependent zinc metallopeptidase for both cytoplasmic and membrane proteins. Plays a role in the quality control of integral membrane proteins."

protein_reward(prediction, reference)
````

````python
import os
import json
import logging
from typing import List, Dict, Any, Optional

logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
logger = logging.getLogger(__name__)

# 定义需要被拦截并特殊处理的伪工具
PSEUDO_TOOLS = {"add_to_graph", "report_key_findings", "go_search_agent"}

def get_full_tools_schema() -> List[Dict[str, Any]]:
    """
    返回全局完整的 Tools Schema。
    包含了化学、生物（Protein）、医学文献（PubMed/PubTator）、疾病（Disease）和药物（Drug）工具。
    精简了 Description 以节省 Token，并严格定义了所有的输入参数。
    """
    return [
        # ==================== 化学与生物工具 ====================
        {
            "type": "function",
            "function": {
                "name": "fetch_molecule_profile",
                "description": "Calculates physicochemical properties, ADMET descriptors, functional groups, NMR peaks, and GHS safety/hazards. Use for virtual screening, oral availability, or evaluating safety risks.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "Short task description. e.g., 'virtual screening', 'check safety'"},
                        "identifier": {"type": "string", "description": "Molecule identifier (SMILES, SELFIES, or common name)."}
                    },
                    "required": ["task_name", "identifier"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "fetch_bioactivity_and_targets",
                "description": "Queries ChEMBL to find a molecule's biological targets, mechanism of action, and clinical trial phase. Use for 'targets', 'receptors', or 'drug approval status'.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "Short description. e.g., 'find targets', 'clinical phase'"},
                        "identifier": {"type": "string", "description": "Molecule identifier (SMILES, SELFIES, or common name)."}
                    },
                    "required": ["task_name", "identifier"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "compute_quantum_mechanics",
                "description": "Calculates quantum chemical properties like LUMO/HOMO energy and Dipole Moment.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "Short description, e.g., 'lumo calculation'"},
                        "smiles": {"type": "string", "description": "The exact SMILES/SELFIES string or name of the molecule."}
                    },
                    "required": ["task_name", "smiles"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "fetch_reaction_details",
                "description": "Fetches empirical chemical reaction data (conditions, yield, reagents, catalysts) from ORD database. Supports forward prediction and retrosynthesis search.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "Short descriptive name, e.g., 'predict yield', 'find solvents'"},
                        "reaction_smiles": {
                            "type": "string", 
                            "description": "Chemical query (SMILES, SELFIES, InChIKey, or IUPAC name). Modes: 1) Full exact match: 'Reactants>>Products'; 2) Forward prediction: 'Reactants>>' or 'Reactants'; 3) Retrosynthesis/Conditions: '>>Product' or 'Product'."
                        },
                        "limit": {"type": "integer", "description": "Maximum number of reaction records to return.", "default": 5}
                    },
                    "required": ["task_name", "reaction_smiles"]
                }
            }
        },
{
            "type": "function",
            "function": {
                "name": "query_protein",
                "description": "Analyzes a protein from sequence or UniProt accession. Retrieves sequence properties, homologs, annotations, catalytic info, pathways, and GO terms.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "Short description used for output filename, e.g., 'analyze protein'"},
                        "sequence": {"type": "string", "description": "Protein amino-acid sequence or FASTA. Provide exactly one of sequence or accession."},
                        "accession": {"type": "string", "description": "UniProt primary accession. Provide exactly one of sequence or accession."},
                        "query": {"type": "string", "description": "Free-form protein question, e.g., 'identify its function'."}
                    },
                    "required": ["task_name"]
                }
            }
        },

        # ==================== PubMed 与 PubTator 工具 ====================
        {
            "type": "function",
            "function": {
                "name": "get_paper_references",
                "description": "Get paper references (citations) for a list of PubMed IDs.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "A less than three word description of what the search is for."},
                        "pmids": {
                            "type": "array",
                            "items": {"type": "string"},
                            "description": "List of PubMed IDs (PMIDs) to get references for."
                        },
                        "batch_size": {"type": "integer", "default": 100},
                        "mini_batch_size": {"type": "integer", "default": 20},
                        "max_workers": {"type": "integer", "default": 4},
                        "rate_limit": {"type": "number", "default": 3.0}
                    },
                    "required": ["task_name", "pmids"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "fetch_paper_content",
                "description": "Fetch and extract specific content from a PubMed paper by PMID using keyword/regex filtering.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "pmid": {"type": "string", "description": "A single PubMed ID (PMID) to fetch content for."},
                        "filter_keywords": {"type": "string", "description": "Keywords or regex pattern to search for and extract from the paper content."}
                    },
                    "required": ["pmid", "filter_keywords"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "fetch_paper_annotations",
                "description": "Fetch biomedical entity annotations from PubTator3 for a list of PubMed IDs.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "pmids": {
                            "type": "array",
                            "items": {"type": "string"},
                            "description": "List of PubMed IDs (PMIDs) to fetch annotations for."
                        },
                        "batch_size": {"type": "integer", "default": 50},
                        "max_retries": {"type": "integer", "default": 3},
                        "max_requests_per_second": {"type": "number", "default": 3.0}
                    },
                    "required": ["pmids"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "find_entities",
                "description": "Find and autocomplete biomedical entity names in the PubTator3 database.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "query_text": {"type": "string", "description": "A single search term (partial entity name) to find entities."},
                        "concept_type": {
                            "type": "string",
                            "enum": ["GENE", "DISEASE", "CHEMICAL", "VARIANT", "SPECIES", "CELLLINE"],
                            "description": "Optional entity concept type."
                        },
                        "limit": {"type": "integer", "default": 10},
                        "max_retries": {"type": "integer", "default": 3},
                        "max_requests_per_second": {"type": "number", "default": 3.0}
                    },
                    "required": ["query_text"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "search_papers",
                "description": "Search PubMed articles using keywords, boolean queries, or relation queries. TIP: Simple keywords often yield better results.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "A less than three word description of what the search is for."},
                        "boolean_query_text": {"type": "string", "description": "Boolean query or simple natural language text."},
                        "relation_query": {"type": "object", "description": "Relation-based query configuration."},
                        "top_k": {"type": "integer", "default": 100, "description": "Number of top results to return."},
                        "max_retries": {"type": "integer", "default": 3},
                        "max_requests_per_second": {"type": "number", "default": 3.0}
                    },
                    "required": ["task_name"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "find_related_entities",
                "description": "Find entities that have a specific relationship with a target entity using PubTator3 relations API.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "target_entity": {"type": "string", "description": "The target entity to find relations for."},
                        "relation_type": {
                            "type": "string",
                            "enum": [
                                "ASSOCIATE", "CAUSE", "COMPARE", "COTREAT", "DRUG_INTERACT",
                                "INHIBIT", "INTERACT", "NEGATIVE_CORRELATE", "POSITIVE_CORRELATE",
                                "PREVENT", "STIMULATE", "TREAT", "ANY"
                            ],
                            "description": "Type of relation to search for."
                        },
                        "related_entity_type": {"type": "string", "description": "Type or ID of related entities to find."},
                        "limit": {"type": "integer", "default": 100},
                        "max_retries": {"type": "integer", "default": 3},
                        "max_requests_per_second": {"type": "number", "default": 3.0}
                    },
                    "required": ["target_entity", "relation_type", "related_entity_type"]
                }
            }
        },

        # ==================== Disease 疾病信息工具 ====================
        {
            "type": "function",
            "function": {
                "name": "unified_disease_search",
                "description": "Search for diseases across multiple authoritative databases (BioThings, KEGG, Open Targets) with a single search term.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "A less than three word description of what the search is for."},
                        "search_term": {"type": "string", "description": "Disease name, condition, or symptoms."},
                        "limit_per_source": {"type": "integer", "default": 10},
                        "sources": {
                            "type": "array",
                            "items": {"type": "string"},
                            "description": "List of sources. Options: 'biothings', 'kegg', 'opentargets'."
                        }
                    },
                    "required": ["task_name", "search_term"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "fetch_disease_details",
                "description": "Fetch comprehensive disease details using any identifier (MONDO, DOID, OMIM, MeSH, KEGG, EFO, or name).",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "A less than three word description of what the fetch is for."},
                        "disease_id": {"type": "string", "description": "Disease identifier or disease name."},
                        "id_type": {"type": "string", "description": "Type of identifier if known. Auto-detected if not specified."},
                        "sources": {
                            "type": "array",
                            "items": {"type": "string"},
                            "description": "List of sources. Options: 'biothings', 'kegg', 'opentargets'."
                        }
                    },
                    "required": ["task_name", "disease_id"]
                }
            }
        },

        # ==================== Drug 药物信息工具 ====================
        {
            "type": "function",
            "function": {
                "name": "unified_drug_search",
                "description": "Search for drugs across multiple databases (BioThings, OpenFDA, KEGG, Open Targets). Returns FDA status, labels, and trials.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "A less than three word description of what the search is for."},
                        "search_term": {"type": "string", "description": "Drug name, compound name, condition, or any search term."},
                        "limit_per_source": {"type": "integer", "default": 10},
                        "sources": {
                            "type": "array",
                            "items": {"type": "string"},
                            "description": "List of sources. Options: 'biothings', 'openfda_approval', 'openfda_label', 'kegg', 'opentargets'."
                        }
                    },
                    "required": ["task_name", "search_term"]
                }
            }
        },
        {
            "type": "function",
            "function": {
                "name": "fetch_drug_details",
                "description": "Fetch comprehensive drug details using any identifier (DrugBank, ChEBI, ChEMBL, PubChem, OpenFDA, KEGG, or name).",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "task_name": {"type": "string", "description": "A less than three word description of what the fetch is for."},
                        "drug_id": {"type": "string", "description": "Drug identifier of any type or name."},
                        "id_type": {"type": "string", "description": "Type of identifier if known. Auto-detected if not specified."},
                        "sources": {
                            "type": "array",
                            "items": {"type": "string"},
                            "description": "List of sources. Options: 'biothings', 'openfda_approval', 'openfda_label', 'kegg', 'opentargets'."
                        }
                    },
                    "required": ["task_name", "drug_id"]
                }
            }
        },
        # ==================== 代码执行工具 ====================
        {
            "type": "function",
            "function": {
                "name": "code_execution",
                "description": "Execute Python code in a secure sandbox environment. Useful for custom calculations, data manipulation, or using specific libraries (e.g., RDKit, SELFIES, pandas) when predefined tools are insufficient.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "code": {
                            "type": "string", 
                            "description": "The Python code script to execute."
                        },
                        "language": {
                            "type": "string", 
                            "description": "The programming language to use (e.g., 'python').", 
                            "default": "python"
                        }
                    },
                    "required": ["code"]
                }
            }
        }
    ]

def process_single_sample(sample_dir: str) -> Optional[Dict[str, Any]]:
    meta_path = os.path.join(sample_dir, "meta.json")
    successful_path_file = os.path.join(sample_dir, "successful_tool_path.json")
    final_response_file = os.path.join(sample_dir, "final_response.md")

    if not os.path.exists(successful_path_file) or not os.path.exists(meta_path):
        return None

    try:
        with open(meta_path, "r", encoding="utf-8") as f:
            meta_data = json.load(f)
        user_query = meta_data.get("user_query") or meta_data.get("question")
        if not user_query: return None

        with open(successful_path_file, "r", encoding="utf-8") as f:
            successful_path = json.load(f)

        final_answer = ""
        if os.path.exists(final_response_file):
            with open(final_response_file, "r", encoding="utf-8") as f:
                final_answer = f.read().strip()
    except Exception as e:
        logger.error(f"读取异常 [{sample_dir}]: {e}")
        return None

    # 获取全局全量工具列表
    tools_schema = get_full_tools_schema()
    
    messages = [
        {"role": "user", "content": user_query}
    ]

    for step in successful_path:
        tool_name = step.get("tool_name", "")
        tool_input = step.get("tool_input", {})
        raw_output = step.get("raw_output")

        # ---------------------------------------------------------
        # 核心清洗逻辑：拦截并处理“伪工具”
        # ---------------------------------------------------------
        
        # 1. 废弃采样工具 (完全忽略)
        if tool_name == "add_to_graph":
            continue
            
        # 2. 将任务下发 (go_search_agent) 转化为内部思考/规划
        elif tool_name == "go_search_agent":
            target = tool_input.get("search_target", "")
            if target:
                thought = f"<think>\nPlan: I need to search biomedical literature for the following target: {target}\n</think>"
                messages.append({"role": "assistant", "content": thought})
            continue # 跳过 tool_response
            
        # 3. 将中间总结 (report_key_findings) 转化为内部反思
        elif tool_name == "report_key_findings":
            findings = tool_input.get("key_findings", [])
            findings_text = "\n".join([f"- {f.get('finding', '')}" for f in findings])
            if findings_text:
                thought = f"<think>\nKey Findings Summary:\n{findings_text}\n</think>"
                messages.append({"role": "assistant", "content": thought})
            continue # 跳过 tool_response

        # ---------------------------------------------------------
        # 正常工具调用逻辑
        # ---------------------------------------------------------
        else:
            # === 新增：强制清理 query_protein 的历史冗余参数 ===
            if tool_name == "query_protein":
                allowed_keys = {"task_name", "sequence", "accession", "query"}
                # 只保留白名单内的参数
                tool_input = {k: v for k, v in tool_input.items() if k in allowed_keys}

            tool_call_content = {
                "name": tool_name,
                "arguments": tool_input
            }
            messages.append({
                "role": "tool_call",
                "content": json.dumps(tool_call_content, ensure_ascii=False)
            })

            if raw_output is None:
                raw_output_str = json.dumps({"status": step.get("status", "success")}, ensure_ascii=False)
            else:
                raw_output_str = raw_output if isinstance(raw_output, str) else json.dumps(raw_output, ensure_ascii=False)
                
            messages.append({
                "role": "tool_response",
                "content": raw_output_str
            })

    # 4. 合并连续的 assistant 消息 (避免格式突兀)
    merged_messages = []
    for msg in messages:
        if merged_messages and merged_messages[-1]["role"] == "assistant" and msg["role"] == "assistant":
            merged_messages[-1]["content"] += "\n" + msg["content"]
        else:
            merged_messages.append(msg)

    # 5. 追加最终回复
    if final_answer:
        if merged_messages[-1]["role"] == "assistant":
            merged_messages[-1]["content"] += "\n" + final_answer
        else:
            merged_messages.append({"role": "assistant", "content": final_answer})

    return {
        "tools": json.dumps(tools_schema, ensure_ascii=False),
        "messages": merged_messages
    }

def main_convert_dataset(root_dir: str, output_file: str):
    if not os.path.exists(root_dir): return
    logger.info("开始单进程递归扫描...")
    
    sample_dirs = []
    for current_root, _, files in os.walk(root_dir):
        if "successful_tool_path.json" in files and "meta.json" in files:
            sample_dirs.append(current_root)

    total_samples = len(sample_dirs)
    logger.info(f"匹配到 {total_samples} 个样本目录")

    with open(output_file, "w", encoding="utf-8") as f:
        for idx, s_dir in enumerate(sample_dirs, 1):
            record = process_single_sample(s_dir)
            if record:
                f.write(json.dumps(record, ensure_ascii=False) + "\n")
            
            if idx % 500 == 0: logger.info(f"已处理 {idx}/{total_samples} ...")

    logger.info(f"数据集构建完成！路径: {output_file}")


if __name__ == "__main__":
    DATASET_ROOT_DIR = "/home/data4/xiangyu/BioDSA/tutorials/workdir"
    OUTPUT_JSONL = "/home/data4/xiangyu/skill-learner/sci_tool_agent_opsd_train.jsonl"
    main_convert_dataset(DATASET_ROOT_DIR, OUTPUT_JSONL)
````

````python
import importlib.util
import json
from pathlib import Path

plugin_path = Path(
    "/home/data4/xiangyu/skill-learner/ms-swift/examples/train/rlhf/opsd/opsd_plugin.py"
)
source_path = Path(
    "/home/data4/xiangyu/skill-learner/sci_tool_agent_opsd_train.jsonl"
)
output_path = Path(
    "/home/data4/xiangyu/skill-learner/sci_tool_agent_opsd_train_processed.jsonl"
)

spec = importlib.util.spec_from_file_location("opsd_plugin_export", plugin_path)
plugin = importlib.util.module_from_spec(spec)
spec.loader.exec_module(plugin)

preprocessor = plugin.BioDSAOPSDPreprocessor()

total = 0
kept = 0

with source_path.open("r", encoding="utf-8") as src, \
     output_path.open("w", encoding="utf-8") as dst:

    for line in src:
        if not line.strip():
            continue

        total += 1
        raw = json.loads(line)
        result = preprocessor.preprocess(raw)

        if result is None:
            continue

        # Keep each record's schema stable and serialize the nested
        # fields as JSON strings where required.
        result["messages"] = json.dumps(
            result["messages"], ensure_ascii=False
        )
        result["tools"] = (
            result["tools"]
            if isinstance(result["tools"], str)
            else json.dumps(result["tools"], ensure_ascii=False)
        )
        result["teacher_prompt"] = str(result["teacher_prompt"])
        result["teacher_tool_trace"] = (
            result["teacher_tool_trace"]
            if isinstance(result["teacher_tool_trace"], str)
            else json.dumps(
                result["teacher_tool_trace"], ensure_ascii=False
            )
        )

        dst.write(json.dumps(result, ensure_ascii=False) + "\n")
        kept += 1

print("Input rows:", total)
print("Processed rows:", kept)
print("Output:", output_path)
print("Output bytes:", output_path.stat().st_size)

# Validate the exported first row.
with output_path.open("r", encoding="utf-8") as f:
    first = json.loads(next(line for line in f if line.strip()))

print("Output keys:", list(first.keys()))
print("Messages:", len(json.loads(first["messages"])))
print("Tools:", len(json.loads(first["tools"])))
print("Trace steps:", len(json.loads(first["teacher_tool_trace"])))

assert kept > 0
assert all(
    k in first
    for k in (
        "messages",
        "tools",
        "teacher_prompt",
        "teacher_tool_trace",
    )
)
print("PREPROCESS EXPORT PASSED")
````

````python
from Bio.Seq import Seq

sequence = "GATTTACCTGATGAGTTCAGAGAAAAACCTTCTGCCCAATGAAGTACAGGCCTATTGCTGGACACTGTTCCGAGTCGCGAGTCCTCGCACAGCCGGCTATTTAAATCGATCCTACCTCTATATTATACAGCCGCCTAGTGAGGACATCGACAGATTAGTGGTTGAATGAAAACGACGTCTTACTAGGTCTCAGTCCTGGCCCGTGGGGCCGTTAGACTATTCCGTCAAGGAAAGAGACCCGTGTAAGCTGTATACTTATTAAATAAATGTCACCACATGGTAGTTACCCACCCACTTCCGGCGGGGGCCGGCGGTCGTCTCCGCGATCGCCCGGATACGGCCGACTTTCAAGGTGTTCTCGTTAAGTGTTCCTTTACGACAATGATAACGCGAACGTGCCCGCATTCAAGAGGTCTCGAAAAATGACGGGACGGTAGGTTCATACTCTCAGAATACGCATACGTTTTAAGCGAGTAAGTTAGATGAGAAGGTACACGAAATATTACAACACAGGGCACGTCACGGCTGATCGGTATGCGTATTTTGCGGAGCTTATACTTAGATTTCGGAAGCCTAACTGGTTTGTCTTTCAGACGGTTCGATGCCCCCCGACTACCCGTACGTGGGGGGCTCGATGTAGCTAGACGTGGAAATTTCATCCTCCTAGCCCGCGTGCTGCCGGTGGGTGAGAAGATAATAAAGTATGGCCTACCCACAAACCGGATACATAACAACCTGGAATTCTCGAACACTCCCTCGTCGCGAACGTGCATTCTGGTTTGTTTTCTCGACATATCCCGACGTATAATTTCAAACCGTAGAGCATACCAAACCAAAGCATCTGGTCGGTCAGTCAAGGGGAAGTTTTGCTCACAACTACCGGCTTTACGTAAATGACGGAATGTAGCATCTCGGACTACGTTACTGCCGTTGCAACCTTTAGGCCGAGGATTACAC"
min_aa_length = 70

# 1. 修正了带有错误大括号的 Seq 实例化操作
seq = Seq(sequence.replace('U', 'T'))
orfs = []

for strand, nuc in [(+1, seq), (-1, seq.reverse_complement())]:
    for frame in range(3):
        # 裁剪掉末尾非 3 的倍数的碱基，避免 Biopython 抛出 Partial Codon 警告
        valid_length = len(nuc) - frame
        valid_length -= valid_length % 3
        valid_nuc = nuc[frame : frame + valid_length]
        
        trans = str(valid_nuc.translate())
        trans_len = len(trans)
        aa_start = 0
        
        while aa_start < trans_len:
            aa_end = trans.find('*', aa_start)
            if aa_end == -1:
                aa_end = trans_len
            
            # 2. 移除了变量周围多余的 { }
            if aa_end - aa_start >= min_aa_length:
                m_start = trans.find('M', aa_start, aa_end)
                if m_start != -1:
                    orf_seq = trans[m_start:aa_end]
                    
                    if len(orf_seq) >= min_aa_length:
                        # 3. 修正了向列表添加字典时多余的双大括号 {{ }}
                        orfs.append({
                            'strand': strand, 
                            'frame': frame, 
                            'length': len(orf_seq), 
                            'aa_seq': orf_seq
                        })
            aa_start = aa_end + 1

# 按长度降序排序
orfs.sort(key=lambda x: x['length'], reverse=True)

# 4. 修正了 f-string 中的变量解析语法，去除了转义用的 {{ }}
print(f"Total ORFs found >= {min_aa_length} AAs: {len(orfs)}")
if orfs:
    longest = orfs[0]
    print(f"Longest ORF length: {longest['length']}")
    print(f"Longest ORF AA sequence: {longest['aa_seq']}")
    if len(longest['aa_seq']) >= 2:
        print(f"Amino acid at position 2 of longest ORF: {longest['aa_seq'][1]}")
````
